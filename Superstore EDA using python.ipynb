```python
import matplotlib.pyplot as plt

category_summary = data.groupby('Category')[['Sales', 'Profit']].sum().reset_index()

fig, ax1 = plt.subplots(figsize=(3, 4))

sns.barplot(data=category_summary, x='Category', y='Sales', ax=ax1)
ax1.set_title('Total Sales by Category')
ax1.set_xlabel('Category')
ax1.set_ylabel('Total Sales')

plt.show()
```

```python
plt.figure(figsize=(3, 4))

sns.barplot(
    data=data,
    x='Category',
    y='Profit',
    estimator='sum',
    errorbar=None
)

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=20)
plt.show()
```

```python
region_sales = data.groupby('Region')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(3, 4))
sns.barplot(x=region_sales.index, y=region_sales.values)

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
region_profit = data.groupby('Region')['Profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(3, 4))
sns.barplot(x=region_profit.index, y=region_profit.values)

plt.title('Total Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.show()
```

```python
# Pie chart: Sales distribution by Category

sales_by_category = data.groupby("Category")["Sales"].sum()

plt.figure(figsize=(3, 3))
plt.pie(
    sales_by_category,
    labels=sales_by_category.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Category")
plt.show()
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])

monthly_sales = (
    data.groupby(data['Order Date'].dt.to_period('M'))['Sales']
    .sum()
    .reset_index()
)

monthly_sales['Order Date'] = monthly_sales['Order Date'].dt.to_timestamp()

plt.figure(figsize=(8, 5))
sns.lineplot(
    data=monthly_sales,
    x='Order Date',
    y='Sales'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Order Date')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_profit = (
    data.groupby(data['Order Date'].dt.to_period('M'))['Profit']
    .sum()
    .reset_index()
)

monthly_profit['Order Date'] = monthly_profit['Order Date'].dt.to_timestamp()

plt.figure(figsize=(8, 5))
sns.lineplot(
    data=monthly_profit,
    x='Order Date',
    y='Profit'
)

plt.axhline(0, linestyle='--')

plt.title('Monthly Profit Trend')
plt.xlabel('Order Date')
plt.ylabel('Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
columns = ['Ship Mode', 'Segment', 'Region', 
           'Category', 'Sub-Category']

# Number of columns for the subplot grid
n_cols = 3
n_rows = (len(columns) + n_cols - 1) // n_cols

# Create subplots
fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(15, n_rows * 5)
)

# Flatten axes array
axes = axes.flatten()

# Iterate over columns and create count plots
for i, col in enumerate(columns):
    sns.countplot(
        x=col,
        data=data,
        ax=axes[i]
    )
    
    axes[i].set_title(f'Count Plot of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')
    
    # Rotate labels for better readability
    axes[i].tick_params(axis='x', rotation=45)

# Remove empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
